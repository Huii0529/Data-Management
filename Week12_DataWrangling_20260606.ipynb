## **4. Binning and outlier handling**
- used to <font color='red'>**transform continuous data**</font> into <font color='red'>**discrete categories**</font> or <font color='red'>**intervals (bins)**</font>

```python
# Mock data
ages = [20, 22, 25, 27, 21, 23, 37, 31, 61, 45, 41, 32]
```

### **Understanding Interval Notation**

In pandas binning output:

- `(18, 25]` means values greater than 18 and up to 25 are included.
- `(` or `)` means the boundary is excluded.
- `[` or `]` means the boundary is included.

For example, `(18, 25]` includes 19, 20, 21, 22, 23, 24, and 25, but excludes 18.

```python
# Divide into bins of 18 to 25, 26 to 35, 36 to 60, and finally 61 and older
# returns a Categorical object
bins = [18, 25, 35, 100]
age_cut = pd.cut(ages, bins)
age_cut
```

### **Unique pandas categorical object**

```python
# Codes
age_cut.codes
```

```python
# Categories
age_cut.categories
```

```python
# Accessing single elements
# closed -> inclusive
age_cut.categories[2]
```

```python
# Count number of each category
pd.Series(age_cut).value_counts()
```

```python
# Method 2: Count the number of observations in each age group
age_cut.value_counts()
```

```python
# Inclusive of left, exclusive (not include) right
pd.cut(ages, bins, right=False)
```

```python
# Exclusive of left, inclusive of right
pd.cut(ages, bins, right=True)
```

```python
# Override the default interval-based bin labeling by passing a list or
# array to the labels option
gp_names = ["Youth", "MiddleAge", "Senior"]
pd.cut(ages, bins, labels=gp_names)
```

```python
# Some mock data
data = np.random.uniform(size=20)
data
```

## **5. Permutation and Random Sampling**
- Permutation - process of rearranging the order of elements within a dataset
- This rearrangement is often **done randomly**, meaning that the elements are shuffled into a **new, unpredictable order**.

```python
# Create a 2D array with 5 rows and 7 columns
df = pd.DataFrame(np.arange(35).reshape(5, 7))
df
```

```python
# Randomly rearrange the row order
np.random.seed(123)

sampler = np.random.permutation(5)
sampler
```

```python
# Reorder the rows based on the random row index order
df.iloc[sampler]
```

```python
# Randomly reorder all rows using sample() with a fixed random state
df.sample(frac=1, random_state=123)
```

### **Notes:**
- **`frac=1`** means sample 100% of the rows
- Since all rows are selected but in random order, the DataFrame is shuffled.

```python
# Generate a random order for the 7 column indexes
column_sampler = np.random.permutation(7)
column_sampler
```

```python
# Randomly reorder the columns of the DataFrame
print(df)
print('\n--------\n')

# Generate a random order for the 7 column indexes
column_sampler = np.random.permutation(7) # changed every time runned the codes

# Reorder the columns based on the random column order
df.iloc[:, column_sampler]
```

## **6. String cleaning and regular expressions**
String data often contains inconsistent spacing, punctuation, and formatting.
- Before text can be analysed, we usually need to split, trim, search, and extract patterns.
- Simple string methods are useful for basic cleaning, while regular expressions are useful when the pattern is more complex.

### **String Manipulation**

```python
# A comma-separated string can be broken into pieces with split
val = "a,b,  abu  bakar"
val.split(",")
```

```python
# Split is often combined with strip to trim whitespace
pieces = [x.strip() for x in val.split(",")]
pieces
```

```python
# Substrings could be concatenated together with addition
first, second, third = pieces
first + "&" + second + "&" + third
```

```python
# Another way
"@".join(pieces)
```

```python
# Return starting index of the first occurrence of passed substring
# If found in the string
val.index("u")
```

```python
# Count returns the number of occurrences
val.count(",")
```

### **Regular Expressions**
***regex*** describes a ***pattern to locate*** in the text, which can then be used for many purposes

### <font color='red'>**`r"\s+"`**</font>
- <font color='red'>**`r""`**</font>: This designates a **raw string literal**, which is often used for regular expressions to avoid escaping issues with **backslashes (`\`)**.
- <font color='red'>**`\s+`**</font>: This is the actual regular expression pattern. It **matches one or more whitespace characters**.

```python
# Regex describing one or more whitespace characters -> \s+
# raw string (r"\s+") is often preferred in regular expressions to enhance readability and avoid potential issues with escaping
import re
text = "foo bar\t baz \tqux"
print(text)
print('\n----\n')
re.split(r"\s+", text)
```

```python
# Compile the regex with re.compile, forming a reusable regex object
# r'\s+' refers to sequence of whitespace
# Avoid unwanted escaping with \ in a regular expression
regex = re.compile(r"\s+")
regex.split(text)
```

```python
# A block of text example
text = """
Dave dave@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@yahoo.com
"""

# Display text
print(text)

# The pattern to search
# Match a typical email address pattern
pattern = r"[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}"

# re.IGNORECASE makes the regex case insensitive
regex = re.compile(pattern, flags=re.IGNORECASE)
```

```python
# Extract the information based on the pattern
regex.findall(text)
```

## **8. Combining datasets: merge, join, concatenate**
After cleaning individual datasets, the next common task is to combine information from multiple tables.
- This is especially important when different parts of the data are stored separately, such as customer information in one table and transaction records in another table.

In pandas, the most common tools are:

- `merge()` for database-style joins using key columns.
- `concat()` for stacking or combining objects along rows or columns.
- `join()` for combining based on indexes.

### **Database-Style DataFrame Joins**

```python
# Example datasets
df1 = pd.DataFrame({"key": ["b", "b", "a", "c", "a", "a", "b"],
                    "data1": pd.Series(range(7), dtype="Int64")})
df2 = pd.DataFrame({"key": ["a", "b", "d"],
                    "data2": pd.Series(range(3), dtype="Int64")})
```

```python
# Display df1 content
df1
```

```python
# Display df2 content
df2
```

```python
# By default - inner join
# keeps only the rows where the "key" values exist in both DataFrames
# Using the same common "feature"
pd.merge(df1, df2)
```

```python
# Different column names
df3 = pd.DataFrame({"lkey": ["b", "b", "a", "c", "a", "a", "b"],
                    "data1": pd.Series(range(7), dtype="Int64")})
df4 = pd.DataFrame({"rkey": ["a", "b", "d"],
                    "data2": pd.Series(range(3), dtype="Int64")})
```

```python
# Display df3 content
df3
```

```python
# Display df4 content
df4
```

```python
# Merge - inner join by default - intersection
# Using specific key
pd.merge(df4, df3, left_on="rkey", right_on="lkey")
```

```python
# Merge - outer join - union
print(df1)
print('\n----\n')
print(df2)
print('\n----\n')
pd.merge(df1, df2, how="outer")
```

```python
# Merge - outer join - union
print(df3)
print('\n----\n')
print(df4)
print('\n----\n')
pd.merge(df3, df4, left_on="lkey", right_on="rkey", how="outer")
```

## **9. Reshaping data: stack, unstack, pivot**
1. Before learning `stack()` and `unstack()`, it is useful to understand hierarchical indexing. Hierarchical indexing allows a Series or DataFrame to have more than one index level. This is useful when working with grouped, multi-dimensional, or panel-like data.

2. Once the data has multiple index levels, we can reshape it using methods such as `stack()` and `unstack()`.

### **Reshaping**
reshape or pivoting - rearranging tabular data

### **Pivoting**
- a specific type of **data reshaping** operation where we:

    - **Transform data from a "long" format to a "wide" format (or vice-versa)**.
    - **Use one or more columns as new column headers**.
    - **Aggregate values in other columns to populate the new structure.**

### ***Reshaping with Hierarchical Indexing***

```python
# Example data
data = pd.DataFrame(np.arange(6).reshape((2, 3)),
                    index=pd.Index(["Ohio", "Colorado"], name="state"),
                    columns=pd.Index(["one", "two", "three"],
                                     name="number"))
data
```

```python
# Using the stack method to pivot the columns into the rows
result = data.stack()
result
```

```python
# Rearrange
result.unstack()
```

### **`Int64`**
- Pandas data type that can handle **nullable integers**.
- **Nullable** means it can **store regular integer values** as well as **missing values** (represented as pd.NA).

```python
# Example
s1 = pd.Series([0, 1, 2, 3], index=["a", "b", "c", "d"], dtype="Int64")
s1
```

```python
# example 2
s2 = pd.Series([4, 5, 6], index=["c", "d", "e"], dtype="Int64")
s2
```

- It allows us to apply string-related methods to every value in a Series.

- 1 = Yes, this category is present
- 0 = No, this category is not present
